#Extraer las notas

```python
#Extrae notas y acordes de todos los archivos MIDI en una carpeta
def extract_notes(midi_folder):

    notes = []
    for file in os.listdir(midi_folder):
        if not file.endswith(".mid"):
            continue
        fpath = midi_folder / file
        if fpath.stat().st_size == 0:
            print(f"Archivo vacío: {file}, se omite.")
            continue

        print("Procesando:", file)
        try:
            midi = converter.parse(fpath)
            parts = instrument.partitionByInstrument(midi)
            elements = parts.parts[0].recurse() if parts else midi.flat.notes
        except Exception as e:
            print(f"Error procesando {file}: {e}")
            continue

        for element in elements:
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))
            elif isinstance(element, chord.Chord):
                notes.append('.'.join(str(n.pitch) for n in element.notes))
    return notes


# Extracción por carpeta
notas_castlevania = extract_notes(midi_castlevania)
```

#Intento de mejorar la extracción de notas

##Extraer solo la melodía

Información que le pedí a chatGPT porque no sé nada de música:
- La melodía casi siempre es la pista con más notas individuales (no acordes).
- O la pista con el rango más alto (melodía tiende a tener notas agudas).
- O la pista con instrumento tipo lead (si está etiquetado).

```python
from music21 import converter, instrument, note, chord
import os

def extract_melody(midi_folder):
    notes = []

    for file in os.listdir(midi_folder):
        if not file.endswith(".mid"):
            continue
        fpath = midi_folder / file

        try:
            midi = converter.parse(fpath)
        except:
            print("Error leyendo", file)
            continue

        parts = instrument.partitionByInstrument(midi)

        if not parts:
            elements = midi.flat.notes
        else:
            candidate = None
            max_notes = -1

            # Elegir pista con más notas individuales (melodía probable)
            for p in parts.parts:
                elems = list(p.recurse().notes)
                single_notes = sum(isinstance(e, note.Note) for e in elems)
                if single_notes > max_notes:
                    max_notes = single_notes
                    candidate = elems

            elements = candidate

        for element in elements:
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))
            elif isinstance(element, chord.Chord):
                notes.append('.'.join(str(n.pitch) for n in element.notes))

    return notes

# Extracción por carpeta
notas_melody_castlevania = extract_melody(midi_castlevania)
```

##Usar todas las pistas (varios instrumentos)

```python
from music21 import converter, instrument, note, chord
import os

def extract_multitrack(midi_folder):
    notes = []

    for file in os.listdir(midi_folder):
        if not file.endswith(".mid"):
            continue
        fpath = midi_folder / file

        try:
            midi = converter.parse(fpath)
        except:
            print("Error leyendo", file)
            continue

        parts = instrument.partitionByInstrument(midi)

        # Si no hay separación de instrumentos → usar todas las notas
        if not parts:
            all_elems = list(midi.flat.notes)
        else:
            # Combinar todas las notas de todas las pistas
            all_elems = []
            for p in parts.parts:
                all_elems.extend(list(p.recurse().notes))

        # Ordenar cronológicamente por offset
        all_elems = sorted(all_elems, key=lambda x: x.offset)

        for element in all_elems:
            if isinstance(element, note.Note):
                notes.append(str(element.pitch))
            elif isinstance(element, chord.Chord):
                notes.append('.'.join(str(n.pitch) for n in element.notes))

    return notes

# Extracción por carpeta
notas_multitrack_castlevania = extract_multitrack(midi_castlevania)
```

Convertir las notas a algo interpretable (Procesar los datos)

```python
def process_and_save(notes, prefix, output_dir):
    unique_notes = sorted(set(notes))
    note_to_int = {n: i for i, n in enumerate(unique_notes)}
    encoded = [note_to_int[n] for n in notes]

    SEQ_LENGTH = 50
    inputs, targets = [], []
    for i in range(len(encoded) - SEQ_LENGTH):
        inputs.append(encoded[i:i+SEQ_LENGTH])
        targets.append(encoded[i+SEQ_LENGTH])

    inputs = np.array(inputs)
    targets = np.array(targets)

    np.save(output_dir / f"{prefix}_notas.npy", np.array(notes))
    np.savez(output_dir / f"{prefix}_secuencias_datos.npz", inputs=inputs, targets=targets)
    json.dump(unique_notes, open(output_dir / f"{prefix}_notas_unicas.json", "w"))

    print(f"Dataset '{prefix}' guardado: {inputs.shape[0]} secuencias, {len(unique_notes)} notas únicas")
```

```python
process_and_save(notas_castlevania, "notas_castlevania", DATA_DIR)
process_and_save(notas_melody_castlevania, "notas_melody_castlevania", DATA_DIR)
process_and_save(notas_multitrack_castlevania, "notas_multitrack_castlevania", DATA_DIR)
```
